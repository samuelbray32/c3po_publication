```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
true_value = False

c_proj = np.array([0.5, 0.5])

xx = np.linspace(-2, 2, 100)
yy = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(xx, yy)
loss = np.zeros_like(X)
for i in range(len(xx)):
    for j in range(len(yy)):
        z_m = np.array([X[i, j], Y[i, j]])
        loss[i, j] = -np.dot(c_proj + z_m, z_m)


plt.figure(figsize=(8, 6))
plt.contourf(
    X,
    Y,
    -loss,
    levels=30,
    cmap="viridis",
    clim=[
        -0.5,
        #  loss.max()
        10,
    ],
)
plt.colorbar(label="Projection Value")
plt.xlabel("Embedded mark space 1")
plt.ylabel("Embedded mark space 2")
plt.title("Contour Plot of $log [H_j / H_j']$")
plt.grid()
plt.scatter(
    c_proj[0], c_proj[1], color="red", label="Projected Context", zorder=10, s=50
)
in_peak = np.unravel_index(np.argmax(loss), loss.shape)
plt.scatter(xx[in_peak[0]], yy[in_peak[1]], color="pink", label="Peak", zorder=10, s=50)

# plt.scatter(
#     c_proj[0] ** 2,
#     c_proj[1] ** 2,
#     color="blue",
#     label="Projection Point Sqrt",
#     zorder=10,
#     s=50,
# )
plt.legend()
plt.show()
```
